```python
import pandas as pd
import numpy as np
import os
```

```python
df = pd.read_csv(r'D:\Hackathon\train1\train_full.csv',sep=';')  
```

```python
assert len(df) == 45211, "The CSV file does not have 45211 rows"  
save_dir = r"D:\Hackathon"  
```

```python
df_train = df.iloc[:40000]  
df_test = df.iloc[40000:]  
df_test_ans = df.iloc[40000:]
df_test_sample = df.iloc[40000:]
```

```python
df_train
```

```python
df_test = df_test.drop('y', axis=1)  
df_test_sample = df_test_sample.drop('y', axis=1)  

np.random.seed(42) # for reproducibility
df_test_sample['y'] = np.random.choice(['yes', 'no'], size=len(df_test_sample))
```

```python
test_path = os.path.join(save_dir, 'test.csv')  
train_path = os.path.join(save_dir, 'train.csv')  
test_ans_path = os.path.join(save_dir , 'test_ans.csv')
test_ans_my = os.path.join(save_dir , 'test_ans_my.csv')



# Save the test and train DataFrames to new CSV files  
df_test.to_csv(test_path, index=False)  
df_train.to_csv(train_path, index=False) 
df_test_ans.to_csv(test_ans_path,index=False) 
df_test_sample.to_csv(test_ans_my,index=False)

# Verify the files were created and print their sizes  
if os.path.exists(test_path) and os.path.exists(train_path):  
    test_size = os.path.getsize(test_path) / (1024 * 1024)  # Size in MB  
    train_size = os.path.getsize(train_path) / (1024 * 1024)  # Size in MB  
    
    print("\nFiles successfully created in", save_dir)  
    print(f"test.csv - Rows: {len(df_test)}, Columns: {len(df_test.columns)}, Size: {test_size:.2f} MB")  
    print(f"train.csv - Rows: {len(df_train)}, Columns: {len(df_train.columns)}, Size: {train_size:.2f} MB")  
else:  
    print("Error: Files were not created successfully")  
```
